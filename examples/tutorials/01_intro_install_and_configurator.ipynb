# 01 · 安装、配置器与第一条训练

**路径：入门 Intro。** 安装 editable 包、认识配置器，再用同一公开 API 完成第一条本地训练。

本页运行的是固定随机种子、CPU、小步数 **smoke**，只验证完整训练与结果链路，不是精度结论。
浏览器中的方法配置器仍是并行入口：它下载可训练 `.py`，不会在浏览器训练或上传数据。
notebook 源文件位于 `examples/tutorials/`，可下载后在仓库根目录重新执行。


```python
from pathlib import Path
import json
import sys

import openpinn as op

ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "pyproject.toml").is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("OpenPINN", op.__version__, "| repo", ROOT.name)
```

```python
# 安装命令（在终端执行）：python -m pip install -e ".[test,docs]"
result = op.solve(
    "poisson", method="baseline", steps=2, seed=0, device="cpu",
    model={"hidden_layers": [8, 8]},
    sampling={"interior": 12, "boundary": 4},
    overrides={"evaluation": {"points": 17}},
    output_dir=Path("/tmp/openpinn-m17-intro-first-run"), plot=False,
)
print(json.dumps({
    "problem": result.metrics["problem"],
    "steps": result.metrics["steps"],
    "final_total_loss": result.metrics["final_total_loss"],
}, ensure_ascii=False, indent=2))
```

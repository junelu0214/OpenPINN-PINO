# 03 · PINN：一维 Poisson

**路径：PINN。** 用普通 PINN 完成 Poisson 的训练、解析参考评估和指标读取。

本页运行的是固定随机种子、CPU、小步数 **smoke**，只验证完整训练与结果链路，不是精度结论。
浏览器中的方法配置器仍是并行入口：它下载可训练 `.py`，不会在浏览器训练或上传数据。
notebook 源文件位于 `examples/tutorials/`，可下载后在仓库根目录重新执行。


```python
from pathlib import Path
import json
import sys

import openpinn as op

ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "pyproject.toml").is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("OpenPINN", op.__version__, "| repo", ROOT.name)
```

```python
result = op.solve(
    "poisson", method="baseline", steps=5, seed=0, device="cpu",
    model={"hidden_layers": [16, 16]},
    sampling={"interior": 24, "boundary": 4},
    overrides={"evaluation": {"points": 33}},
    output_dir=Path("/tmp/openpinn-m17-pinn-poisson"), plot=False,
)
print(json.dumps({key: result.metrics[key] for key in (
    "final_total_loss", "pde_residual_rmse", "relative_l2_error", "boundary_rmse"
)}, indent=2))
```

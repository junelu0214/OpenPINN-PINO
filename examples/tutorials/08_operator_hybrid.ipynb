# 二维 Poisson 参数族的混合监督

**问题：** 学习二维 Poisson 参数族的 forcing-to-solution 映射；FNO 与 `hybrid` 用于同时记录数据项和物理项。

本页运行的是固定随机种子、CPU、小步数 **smoke**，只验证完整训练与结果链路，不是精度结论。
浏览器中的方法配置器仍是并行入口：它下载可训练 `.py`，不会在浏览器训练或上传数据。
notebook 源文件位于 `examples/tutorials/`，可下载后在仓库根目录重新执行。


```python
from pathlib import Path
import json
import sys

import openpinn as op

ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "pyproject.toml").is_file())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("OpenPINN", op.__version__, "| repo", ROOT.name)
```

```python
result = op.train_operator(
    "fno", training_mode="hybrid", steps=8, seed=0, device="cpu",
    width=4, n_layers=1, n_modes=2, project_width=8,
    grid_size=16, batch_size=2, n_harmonics=1, lr=1e-2,
    data_weight=1.0, physics_weight=0.1, data_pretrain_steps=2,
)
print(json.dumps({
    "model": result.operator,
    "training_mode": result.config["training_mode"],
    "initial_data_loss": result.metrics["initial_data_loss"],
    "final_data_loss": result.metrics["final_data_loss"],
    "initial_physics_loss": result.metrics["initial_physics_loss"],
    "final_physics_loss": result.metrics["final_physics_loss"],
}, indent=2))
```
